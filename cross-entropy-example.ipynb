```python
import torch
import torch.nn as nn

loss_fn  = nn.CrossEntropyLoss()

# we have the following categories / classses
# 0 = cat
# 1 = dog
# 2 = bird

# Suppose the correct answer in our case is the dog so
target = torch.tensor([1])

# if our model has ouput the following logits (we have a batch size of just one here)
logits = torch.tensor([2.0, 4.0, 1.0])

# so we have the values (high means more confidence)
# cat = 2.0
# dog = 4.0
# bird = 1.0

# CrossEntropy converts these values to probabilities using a soft max function
# exp(logit) / sum(exp(all_logits))

# Cross entropy only cares about the correct class (dog in this case)
# probabilty for dog = 0.844 using the values above
# then loss = -log(0.844) = 0.17 (low loss). This would be higher if the probability was lower.

loss = loss_fn(logits, target)

print(loss)
```
