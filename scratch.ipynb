```python
import torch
import torch.nn as nn
from torch.nn import functional as F

# bacth size = 4
# block size (number of tokens) = 8
# the number of embeddings which in this example is also the embedding dimension = 65
B, T, C = 4, 8, 65

# create a 65x65 embedding table. So each token gets mapped to a vector with 65 components
embedding_table = nn.Embedding(num_embeddings=C, embedding_dim=C)

# generate a random input size 4 x 8 with random ints between 0 and 64
input = torch.randint(C, (B, T))

print("input")
print(input.shape)
print(input)

# generate some random targets, again size 4 x 8 with random ints between 0 and 64
targets = torch.randint(C, (B, T))

print("targets")
print(targets.shape)
print(targets)

# our output is the embedding table apllied to the input, what does this mean?
# for each batch and each token the integer value is mapped to 65 component vector
# so the output has shape 4 x 8 x 65
# We can think of the input as a fibre space with base space 4 x 8 and a single integer at each point
# after applying the embedding table the fibre space has the same base space but a 65 component vector attached at each point
output = embedding_table(input)

print("output")
print(output.shape)
print(output)

# Our logits are taken as the ouput with the base space re-indexed
# the base space is re-indexed from 4 X 8 to 32
# so we have just moved a rectangle of elements out onto a straight line
# but the 'contents' the vectors have not changed. So logits shape is 32 x 65
logits = output.view(B * T, C)

print("logits")
print(logits.shape)
print(logits)

# our targets are also re-indexed in the same way
# we started with 4 x 8 x 1 and we reshape this as 32 x 1, so shape will be 32
# again shifting the base space from a rectangle of points to a line of points
targets = targets.view(B * T)

print("targets")
print(targets.shape)
print(targets)

# finally we apply the cross entropy with inputs our logits and targets our targets
# looking at the dimensions we have cross_entropy(32 x 65, 32)
loss = F.cross_entropy(logits, targets)

print("loss")
print(loss)
```
